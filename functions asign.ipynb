```python
10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
-In Python, map(), reduce(), and filter() are functional programming tools that operate on iterable data structures like lists. Here's a breakdown of what each function does and how they differ:
1. map()
* Purpose: Applies a function to every item in an iterable and returns a new iterable (usually a map object, which can be converted to a list).
* Syntax: map(function, iterable)
* Example:

numbers = \[1, 2, 3, 4]
squared = list(map(lambda x: x\*\*2, numbers))

# squared => \[1, 4, 9, 16]

2. filter()

* Purpose: Filters items in an iterable based on a function that returns True or False.
* Syntax: filter(function, iterable)
* Example:

numbers = \[1, 2, 3, 4]
even = list(filter(lambda x: x % 2 == 0, numbers))

# even => \[2, 4]

3. reduce()

* Purpose: Applies a function cumulatively to the items of an iterable, reducing it to a single value.
* Syntax: reduce(function, iterable\[, initializer])
* Note: reduce is not a built-in function in Python 3; it needs to be imported from functools.
* Example:

from functools import reduce
numbers = \[1, 2, 3, 4]
product = reduce(lambda x, y: x \* y, numbers)

# product => 24

Summary of Differences:

* map() transforms each element.
* filter() selects elements based on a condition.
* reduce() aggregates all elements into a single value.

Each serves a different purpose but can be combined for powerful data processing pipelines.

```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    """Check if a number is a prime number (between 1 and 200)."""
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
# Test the function for numbers from 1 to 200
for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is a prime number.")
```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
#terms.
class FibonacciIterator:
    def __init__(self, num_terms):
        self.num_terms = num_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.num_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a

# Example usage:
fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

```python
#6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
def powers_of_two(max_exponent):
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent
for power in powers_of_two(5):
    print(power)
```

```python
#7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_line_by_line(file_path):
    """
    Generator function that reads a file line by line.

    Args:
        file_path (str): Path to the file.

    Yields:
        str: The next line from the file.
    """
    with open(file_path, 'r') as file:
        for line in file:
            yield line.rstrip('\n')  # Removes trailing newline character
```

```python
file_path = 'example.txt'

for line in read_file_line_by_line(file_path):
    print(line)
```
